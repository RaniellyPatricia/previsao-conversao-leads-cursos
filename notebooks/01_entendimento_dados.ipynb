# 01 - Entendimento inicial da base de leads

Este notebook faz parte do MVP 1 do projeto de Produto de Dados para Análise e Previsão de Conversão de Leads em Cursos.

O objetivo desta etapa é importar a base de dados no Google Colab e realizar uma primeira verificação estrutural, analisando:

- quantidade de linhas e colunas;
- nomes das colunas;
- tipos de dados;
- valores nulos;
- primeiras linhas da base;
- variável de conversão.

Essa etapa é importante para entender se a base está adequada para as próximas fases de limpeza, análise exploratória e construção do dashboard.


```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files
files.upload = files.upload()
```

```python
df = pd.read_csv("Leads_X_Education_traduzido_PTBR.csv")
```

```python
df.head()
```

## Dimensão da base

Nesta etapa, foi verificada a quantidade de linhas e colunas da base. Essa informação ajuda a entender o tamanho do conjunto de dados analisado.

```python
linhas, colunas = df.shape

print(f"A base possui {linhas} linhas e {colunas} colunas.")
```

## Nomes das colunas

Nesta etapa, foram listadas todas as colunas disponíveis na base para entender quais informações existem sobre os leads.

```python
for coluna in df.columns:
    print(coluna)
```

## Tipos de dados

Nesta etapa, foi verificado o tipo de dado de cada coluna e a quantidade de valores preenchidos. Essa análise ajuda a identificar colunas numéricas, categóricas e possíveis valores ausentes.

```python
df.info()
```

## Diagnóstico de valores nulos

Nesta etapa, foi criada uma tabela para identificar a quantidade e o percentual de valores nulos em cada coluna.

```python
tipos_dados = pd.DataFrame({
    "coluna": df.columns,
    "tipo_dado": df.dtypes.values,
    "valores_preenchidos": df.notnull().sum().values,
    "valores_nulos": df.isnull().sum().values,
    "percentual_nulos": (df.isnull().mean().values * 100).round(2)
})

tipos_dados.sort_values(by="percentual_nulos", ascending=False)
```

## Verificação de linhas duplicadas

Nesta etapa, foi verificado se existem registros duplicados na base, pois duplicidades poderiam distorcer indicadores como total de leads e taxa de conversão.

```python
duplicadas = df.duplicated().sum()

print(f"A base possui {duplicadas} linhas duplicadas.")
```

## Análise da variável de conversão

Nesta etapa, foi analisada a variável `Convertido`, que indica se o lead realizou conversão ou não.

Nesta base:

- `1` representa lead convertido;
- `0` representa lead não convertido.

```python
conversao = df["Convertido"].value_counts().reset_index()
conversao.columns = ["status_conversao", "quantidade"]

conversao["percentual"] = (
    df["Convertido"].value_counts(normalize=True).values * 100
).round(2)

conversao
```

```python
taxa_conversao = df["Convertido"].mean() * 100

print(f"Taxa geral de conversão: {taxa_conversao:.2f}%")
```

## Separação entre colunas numéricas e categóricas

Nesta etapa, as colunas foram separadas conforme o tipo de dado. Essa separação ajuda a definir quais análises fazem sentido para cada variável.

As colunas numéricas serão usadas para análises de médias, distribuição e comportamento dos leads. Já as colunas categóricas serão usadas para contagens, agrupamentos e comparação de conversão entre grupos.

```python
colunas_numericas = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
colunas_categoricas = df.select_dtypes(include=["object"]).columns.tolist()

print("Colunas numéricas:")
print(colunas_numericas)

print("\nColunas categóricas:")
print(colunas_categoricas)
```

## Análise das categorias mais frequentes

Nesta etapa, foram analisados os valores mais frequentes das colunas categóricas. O objetivo é entender quais categorias aparecem com mais frequência na base e identificar colunas úteis para o dashboard do MVP 1.

```python
for coluna in colunas_categoricas:
    print(f"\nColuna: {coluna}")
    print(df[coluna].value_counts().head(10))
```

## Interpretação inicial das categorias

A análise das categorias mais frequentes mostrou que algumas variáveis possuem maior potencial para o MVP 1, principalmente aquelas relacionadas à origem, fonte, atividade, perfil e qualidade do lead.

As colunas `Origem do Lead`, `Fonte do Lead`, `Última Atividade`, `Última Atividade Relevante`, `Especialização`, `Ocupação Atual`, `Qualidade do Lead` e `Perfil do Lead` foram consideradas relevantes para as próximas análises, pois ajudam a entender de onde os leads vêm, como interagem com a empresa e quais perfis têm maior potencial de conversão.

Também foram identificadas colunas com baixo valor analítico para o dashboard inicial, principalmente aquelas com apenas uma categoria ou com predominância quase total de uma única resposta, como `Revista`, `Receber Mais Atualizações Sobre Nossos Cursos`, `Atualizar Sobre Conteúdo de Supply Chain`, `Receber Atualizações Sobre Conteúdo de Marketing Digital` e `Concordo em Pagar o Valor por Cheque`.

A coluna `ID do Prospect` será mantida apenas como identificador, pois possui um valor único para cada lead e não é adequada para gráficos ou análises agregadas.

As colunas com muitos registros como `Não informado` serão analisadas com cuidado nas próximas etapas, pois podem indicar ausência de preenchimento ou limitação na coleta dos dados.
